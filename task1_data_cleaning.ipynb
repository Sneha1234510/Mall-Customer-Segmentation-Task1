# Task 1: Data Cleaning and Preprocessing
Mall Customer Segmentation Data

**Important:** Included CSV is a 10-row demo. Download the full Kaggle CSV into `data/Mall_Customers.csv` for final results.

## 1. Import libraries

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path('../data')
INPUT = DATA_DIR / 'Mall_Customers.csv'
if not INPUT.exists():
    INPUT = DATA_DIR / 'mall_customers_sample.csv'
    print('Using 10-row demonstration sample. Replace with full Mall_Customers.csv for final submission.')
df = pd.read_csv(INPUT)
df.head()
```

## 2. Inspect dataset

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
df.describe(include='all')
```

## 3. Standardize column names and text

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(r'[^a-z0-9]+', '_', regex=True).str.strip('_'))
if 'gender' in df.columns:
    df['gender'] = df['gender'].astype('string').str.strip().str.title().replace({'M':'Male','F':'Female'})
df.head()
```

## 4. Convert data types and remove duplicates

```python
for col in ['customerid', 'age', 'annual_income_k', 'spending_score_1_100']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.drop_duplicates()
print('Shape after duplicate removal:', df.shape)
```

## 5. Handle missing values

```python
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        if df[col].isna().any():
            df[col] = df[col].fillna(df[col].median())
    elif df[col].isna().any():
        modes = df[col].mode(dropna=True)
        if not modes.empty:
            df[col] = df[col].fillna(modes.iloc[0])
if 'customerid' in df.columns:
    df = df.dropna(subset=['customerid'])
    df['customerid'] = df['customerid'].astype('int64')
print('Remaining missing cells:', int(df.isna().sum().sum()))
df.head()
```

## 6. Export cleaned data

```python
output = DATA_DIR / 'mall_customers_cleaned_from_notebook.csv'
df.to_csv(output, index=False)
print('Saved:', output.resolve())
df.info()
```

## 7. Conclusion
